```python

def is_variable(x):
    return isinstance(x, str) and x.islower() and x not in ["apple", "riya", "john"]


def unify(x, y, substitution=None):
    if substitution is None:
        substitution = {}

    if substitution is None:
        return None

    # Apply existing substitutions
    if x in substitution:
        x = substitution[x]

    if y in substitution:
        y = substitution[y]

    # If both expressions are equal
    if x == y:
        return substitution

    # If x is a variable
    if is_variable(x):
        substitution[x] = y
        return substitution

    # If y is a variable
    if is_variable(y):
        substitution[y] = x
        return substitution

    # If both are compound expressions
    if isinstance(x, tuple) and isinstance(y, tuple):

        # Check predicate name and number of arguments
        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            substitution = unify(a, b, substitution)

            if substitution is None:
                return None

        return substitution

    return None


# Example
expr1 = ("Eats", "x", "Apple")
expr2 = ("Eats", "Riya", "y")

result = unify(expr1, expr2)

if result is not None:
    print("Unification Successful!")
    print("Substitution:", result)
else:
    print("Unification Failed!")
```
