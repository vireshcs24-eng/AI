```python

import heapq

# Initial state
initial = [[1, 2, 3],
           [4, 0, 6],
           [7, 5, 8]]

# Goal state
goal = [[1, 2, 3],
        [4, 5, 6],
        [7, 8, 0]]


def heuristic(state):
    count = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != goal[i][j]:
                count += 1
    return count


def astar(initial):
    start = tuple(sum(initial, []))
    target = tuple(sum(goal, []))

    pq = [(heuristic(initial), 0, start)]
    visited = set()
    parent = {start: None}
    cost = {start: 0}

    while pq:
        f, g, state = heapq.heappop(pq)

        if state in visited:
            continue
        visited.add(state)

        if state == target:
            path = []
            while state is not None:
                path.append(state)
                state = parent[state]
            path.reverse()

            for s in path:
                for i in range(0, 9, 3):
                    print(s[i:i+3])
                print()
            print("Moves:", len(path) - 1)
            return

        zero = state.index(0)
        r, c = divmod(zero, 3)

        for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nr, nc = r + dr, c + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                pos = nr * 3 + nc
                new_state = list(state)
                new_state[zero], new_state[pos] = new_state[pos], new_state[zero]
                new_state = tuple(new_state)
                new_g = g + 1

                if new_g < cost.get(new_state, float("inf")):
                    cost[new_state] = new_g
                    parent[new_state] = state
                    h = heuristic([list(new_state[i:i+3]) for i in range(0, 9, 3)])
                    heapq.heappush(pq, (new_g + h, new_g, new_state))


print("A* - Misplaced Tiles")
astar(initial)
```
