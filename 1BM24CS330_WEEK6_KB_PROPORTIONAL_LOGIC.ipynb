```python

# Truth Table Enumeration for Propositional Logic

# Knowledge Base:
# Q -> P
# P -> NOT Q
# Q OR R

symbols = ['P', 'Q', 'R']


# Check whether KB is TRUE in a model
def PL_TRUE_KB(model):

    s1 = (not model['Q']) or model['P']
    s2 = (not model['P']) or (not model['Q'])
    s3 = model['Q'] or model['R']

    return s1 and s2 and s3


# Check query in a model
def PL_TRUE(alpha, model):
    return alpha(model)


# Truth Table Checking Algorithm
def TT_CHECK_ALL(alpha, symbols, model):

    if len(symbols) == 0:

        if PL_TRUE_KB(model):
            return PL_TRUE(alpha, model)

        else:
            return True

    else:

        P = symbols[0]
        rest = symbols[1:]

        model_true = model.copy()
        model_true[P] = True

        model_false = model.copy()
        model_false[P] = False

        return (TT_CHECK_ALL(alpha, rest, model_true)
                and
                TT_CHECK_ALL(alpha, rest, model_false))


# Main Entailment Function
def TT_ENTAILS(alpha):

    return TT_CHECK_ALL(alpha, symbols, {})


# Display Truth Table
print("P Q R | Q->P | P->~Q | QvR | KB")

for p in [True, False]:
    for q in [True, False]:
        for r in [True, False]:

            model = {'P': p, 'Q': q, 'R': r}

            s1 = (not q) or p
            s2 = (not p) or (not q)
            s3 = q or r

            kb = s1 and s2 and s3

            print(p, q, r, "|", s1, "|", s2, "|", s3, "|", kb)


# Define Queries
queries = {
    "R": lambda m: m['R'],

    "R -> P": lambda m: (not m['R']) or m['P'],

    "Q -> R": lambda m: (not m['Q']) or m['R']
}


# Check Entailment
print("\nENTAILMENT RESULTS")

for query, expression in queries.items():

    result = TT_ENTAILS(expression)

    print(query, ":", result)
```
