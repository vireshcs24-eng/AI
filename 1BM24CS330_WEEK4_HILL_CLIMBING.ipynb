```python

N = 4


def cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if (state[i] == state[j] or
                abs(state[i] - state[j]) == abs(i - j)):
                conflicts += 1

    return conflicts


def get_neighbors(state):
    neighbors = []

    for i in range(N):
        for j in range(i + 1, N):
            new_state = state.copy()

            new_state[i], new_state[j] = new_state[j], new_state[i]

            neighbors.append((new_state, (i + 1, j + 1)))

    return neighbors


def hill_climbing():
    state = [3, 1, 2, 0]

    while True:
        current_cost = cost(state)

        print("\nCurrent state:", state)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("Solution found!")
            return

        neighbors = get_neighbors(state)

        evaluated = []

        for neighbor, columns in neighbors:
            c = cost(neighbor)
            evaluated.append((neighbor, columns, c))

            print("Neighbor:", neighbor,
                  "Swapped columns:", columns,
                  "Cost:", c)

        min_cost = min(item[2] for item in evaluated)


        best = min(
            (item for item in evaluated
             if item[2] == min_cost),
            key=lambda item: item[1]
        )

        next_state, chosen_columns, next_cost = best

        print("Minimum cost:", min_cost)
        print("Chosen columns:", chosen_columns)
        print("Next state:", next_state)

        if next_cost >= current_cost:
            print("No better neighbour. Stop.")
            return

        state = next_state


hill_climbing()

```
