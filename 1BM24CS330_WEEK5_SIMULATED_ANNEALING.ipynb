```python

import random
import math

N = 8

# Count attacking pairs
def cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if (state[i] == state[j] or
                abs(state[i] - state[j]) == abs(i - j)):
                conflicts += 1

    return conflicts


# Change the row of one queen
def get_neighbor(state):
    neighbor = state.copy()
    col = random.randrange(N)
    new_row = random.randrange(N)

    while new_row == neighbor[col]:
        new_row = random.randrange(N)

    neighbor[col] = new_row
    return neighbor


def simulated_annealing():
    state = [1] * N
    iterations = 0

    while True:
        temperature = 10
        current_state = state.copy()

        while temperature > 0.001:
            current_cost = cost(current_state)

            if current_cost == 0:
                print("Solution found:", current_state)
                print("Total iterations:", iterations)
                return

            neighbor = get_neighbor(current_state)
            next_cost = cost(neighbor)
            iterations += 1

            delta = next_cost - current_cost

            if (delta < 0 or
                random.random() < math.exp(-delta / temperature)):
                current_state = neighbor

            temperature *= 0.90

        # Restart from the original state if unsuccessful
        state = [1] * N


simulated_annealing()
```
